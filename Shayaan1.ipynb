```python
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

print("English-to-Hindi model loaded successfully!")
```

```python
def translate_english_to_hindi(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=512
    )

    translated = model.generate(
        **inputs,
        max_length=512,
        num_beams=4
    )

    hindi_text = tokenizer.decode(
        translated[0],
        skip_special_tokens=True
    )

    return hindi_text
```

```python
english_text = "Hello, how are you?"

hindi_text = translate_english_to_hindi(english_text)

print("English:", english_text)
print("Hindi:", hindi_text)
```

```python
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

print("Model loaded successfully!")
```

```python
def translate_english_to_hindi(text):
    if not text or not text.strip():
        return "Please enter English text."

    inputs = tokenizer(
        text.strip(),
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        **inputs,
        max_length=512,
        num_beams=4
    )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
```

```python
import gradio as gr

demo = gr.Interface(
    fn=translate_english_to_hindi,
    inputs=gr.Textbox(
        lines=5,
        label="English Input",
        placeholder="Enter English text here..."
    ),
    outputs=gr.Textbox(
        lines=5,
        label="Hindi Translation"
    ),
    title="English to Hindi Translator",
    description="Powered by Hugging Face Transformers",
    examples=[
        ["Hello, how are you?"],
        ["India is a beautiful country."],
        ["Machine learning is very interesting."]
    ],
    submit_btn="Translate",
    clear_btn="Clear"
)

demo.launch(share=True)
```
